```python
# Task 2: EDA on Titanic Dataset
## 1. Meaningful Questions
- Which gender had a higher survival rate?
- Did passenger class affect survival?
- Did age affect survival?
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)
df.head()
```

```python
##2.Data Structure
```

```python
print(df.shape)
df.info()
```

```python
## TRENDS,PATTERNS, ANOMALIES
```

```python
print(df.groupby('Sex')['Survived'].mean())
print(df.groupby('Pclass')['Survived'].mean())

sns.countplot(x='Survived', hue='Sex', data=df)
plt.show()
```

```python
sns.boxplot(x=df['Fare'])
plt.show()
```

```python
##! HYPOTHESIS TESTING
```

```python
print(df.groupby('Sex')['Survived'].mean())

sns.heatmap(df.corr(numeric_only=True), annot=True, cmap='coolwarm')
plt.show()
```

```python
## DATA ISSUES 📈
```

```python
print(df.isnull().sum())
print("Duplicates:", df.duplicated().sum())
```

```python
##CONCLUSION ✨
```

```python
## Conclusion
The EDA shows that females and first-class passengers had a much higher survival rate than males and third-class passengers. Fare has a few extreme outliers, and Age and Cabin contain missing values that need to be handled before further analysis. Overall, gender and passenger class were the strongest factors affecting survival.
```
